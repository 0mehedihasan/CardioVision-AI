```python
# ============================================================
# CARDIOVISION AI
# CAMUS ECHO SEGMENTATION
# COMPLETE ONE-CELL PIPELINE (FIXED FOR KAGGLE NOTEBOOK)
#
# FIX SUMMARY:
#   1) NUM_WORKERS = 0 to avoid DataLoader worker pickling issues.
#   2) XAI now uses gradient-based saliency (no hooks) to avoid
#      KeyError caused by missing hook activations.
# ============================================================


import os
import sys
import random
import warnings
import subprocess

os.environ.setdefault("CUDA_VISIBLE_DEVICES", "0")

warnings.filterwarnings("ignore")

required_packages = {
    "nibabel": "nibabel",
    "segmentation_models_pytorch": "segmentation-models-pytorch",
}

for module_name, package_name in required_packages.items():
    try:
        __import__(module_name)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package_name])


from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import nibabel as nib
import matplotlib.pyplot as plt

from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split

import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
import segmentation_models_pytorch as smp


# ============================================================
# CONFIGURATION
# ============================================================

DATASET_ROOT = Path(
    "/kaggle/input/datasets/bubtresearchwork/"
    "camus-extracted/camus/database_nifti"
)

OUTPUT_DIR = Path("/kaggle/working/cardioVision_CAMUS")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PATH = OUTPUT_DIR / "cardiovision_echo_unetplusplus_best.pth"
LAST_CHECKPOINT_PATH = OUTPUT_DIR / "cardiovision_echo_unetplusplus_last.pth"

SEED = 42
IMAGE_SIZE = 256
NUM_CLASSES = 4
BATCH_SIZE = 8
EPOCHS = 50
LEARNING_RATE = 2e-4
WEIGHT_DECAY = 1e-4
NUM_WORKERS = 0          # <-- KEY FIX: no DataLoader workers
PATIENCE = 8


# ============================================================
# GPU CHECK
# ============================================================

print("=" * 70)
print("KAGGLE GPU CONFIGURATION")
print("=" * 70)

if not torch.cuda.is_available():
    raise RuntimeError("CUDA GPU is NOT available. Enable GPU T4 in Kaggle settings.")

DEVICE = torch.device("cuda")
GPU_NAME = torch.cuda.get_device_name(0)

print("CUDA available :", torch.cuda.is_available())
print("GPU            :", GPU_NAME)
print("CUDA version   :", torch.version.cuda)
print("GPU count      :", torch.cuda.device_count())
print("Visible devices:", os.environ.get("CUDA_VISIBLE_DEVICES"))

if "T4" not in GPU_NAME:
    print("\nWARNING: Expected NVIDIA T4, detected:", GPU_NAME)
else:
    print("✓ NVIDIA T4 detected")


# ============================================================
# REPRODUCIBILITY
# ============================================================

def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)
    torch.backends.cudnn.benchmark = True

seed_everything(SEED)


# ============================================================
# DATASET VALIDATION & INDEX
# ============================================================

print("\n" + "=" * 70)
print("CAMUS DATASET")
print("=" * 70)

if not DATASET_ROOT.exists():
    raise FileNotFoundError(f"Dataset not found: {DATASET_ROOT}")

patients = sorted([p for p in DATASET_ROOT.iterdir() if p.is_dir() and p.name.startswith("patient")])
print("Patients found:", len(patients))
assert len(patients) == 500, f"Expected 500 patients, found {len(patients)}"

records = []
for patient_dir in tqdm(patients, desc="Indexing patients"):
    pid = patient_dir.name
    for view in ["2CH", "4CH"]:
        for phase in ["ED", "ES"]:
            image_path = patient_dir / f"{pid}_{view}_{phase}.nii"
            mask_path = patient_dir / f"{pid}_{view}_{phase}_gt.nii"
            if image_path.exists() and mask_path.exists():
                records.append({
                    "patient": pid,
                    "view": view,
                    "phase": phase,
                    "image": str(image_path),
                    "mask": str(mask_path)
                })

df = pd.DataFrame(records)
print("\nAnnotated image-mask pairs:", len(df))
assert len(df) == 2000, f"Expected 2000 pairs, found {len(df)}"


# ============================================================
# MASK VALIDATION & DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("MASK VALIDATION")
print("=" * 70)

expected_labels = {0, 1, 2, 3}
label_counts = Counter()
invalid_masks = []

for mask_path in tqdm(df["mask"], desc="Validating masks"):
    mask = nib.load(mask_path).get_fdata().astype(np.uint8)
    labels = np.unique(mask)
    labels_set = set(labels.tolist())
    if not labels_set.issubset(expected_labels):
        invalid_masks.append((mask_path, sorted(labels_set)))
    for label in labels:
        label_counts[int(label)] += np.sum(mask == label)

print("\nDetected labels:", sorted(label_counts.keys()))
print("Invalid masks:", len(invalid_masks))
assert set(label_counts.keys()).issubset(expected_labels)
assert len(invalid_masks) == 0
print("✓ All masks contain only labels 0, 1, 2, 3.")

total_pixels = sum(label_counts.values())
print("\n" + "=" * 70)
print("MASK LABEL DISTRIBUTION")
print("=" * 70)
for label in sorted(label_counts):
    count = label_counts[label]
    percentage = 100 * count / total_pixels
    print(f"Label {label}: {count:,} pixels ({percentage:.4f}%)")


# ============================================================
# PATIENT-LEVEL SPLIT
# ============================================================

print("\n" + "=" * 70)
print("PATIENT-LEVEL SPLIT")
print("=" * 70)

patient_ids = np.array(sorted(df["patient"].unique()))
train_patients, temp_patients = train_test_split(patient_ids, test_size=0.30, random_state=SEED, shuffle=True)
val_patients, test_patients = train_test_split(temp_patients, test_size=0.50, random_state=SEED, shuffle=True)
train_patients = sorted(train_patients)
val_patients = sorted(val_patients)
test_patients = sorted(test_patients)

train_df = df[df["patient"].isin(train_patients)].reset_index(drop=True)
val_df = df[df["patient"].isin(val_patients)].reset_index(drop=True)
test_df = df[df["patient"].isin(test_patients)].reset_index(drop=True)

print(f"Train      : {len(train_patients)} patients | {len(train_df)} pairs")
print(f"Validation : {len(val_patients)} patients | {len(val_df)} pairs")
print(f"Test       : {len(test_patients)} patients | {len(test_df)} pairs")

# Leakage check
assert set(train_patients).isdisjoint(val_patients)
assert set(train_patients).isdisjoint(test_patients)
assert set(val_patients).isdisjoint(test_patients)
print("\n✓ No patient leakage detected.")


# ============================================================
# PREPROCESSING
# ============================================================

def preprocess_image(image):
    image = image.astype(np.float32)
    low = np.percentile(image, 1)
    high = np.percentile(image, 99)
    image = np.clip(image, low, high)
    image = (image - image.min()) / (image.max() - image.min() + 1e-8)
    tensor = torch.from_numpy(image).float()
    tensor = tensor[None, None, ...]
    tensor = F.interpolate(tensor, size=(IMAGE_SIZE, IMAGE_SIZE), mode="bilinear", align_corners=False)
    return tensor[0]

def preprocess_mask(mask):
    mask = mask.astype(np.uint8)
    tensor = torch.from_numpy(mask).float()
    tensor = tensor[None, None, ...]
    tensor = F.interpolate(tensor, size=(IMAGE_SIZE, IMAGE_SIZE), mode="nearest")
    return tensor[0, 0].long()


# ============================================================
# DATASET CLASS
# ============================================================

class CAMUSDataset(Dataset):
    def __init__(self, dataframe):
        self.df = dataframe.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]
        image = nib.load(row["image"]).get_fdata()
        mask = nib.load(row["mask"]).get_fdata()
        image = preprocess_image(image)
        mask = preprocess_mask(mask)
        return {
            "image": image,
            "mask": mask,
            "patient": row["patient"],
            "view": row["view"],
            "phase": row["phase"]
        }


# ============================================================
# DATASETS & DATALOADERS (num_workers=0)
# ============================================================

train_dataset = CAMUSDataset(train_df)
val_dataset = CAMUSDataset(val_df)
test_dataset = CAMUSDataset(test_df)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=NUM_WORKERS, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True)

print("\n" + "=" * 70)
print("DATALOADERS")
print("=" * 70)
print("Train batches:", len(train_loader))
print("Val batches  :", len(val_loader))
print("Test batches :", len(test_loader))
print("Worker start method: n/a (num_workers=0)")


# ============================================================
# MODEL
# ============================================================

print("\n" + "=" * 70)
print("BUILDING MODEL")
print("=" * 70)

model = smp.UnetPlusPlus(
    encoder_name="timm-efficientnet-b3",
    encoder_weights="imagenet",
    in_channels=1,
    classes=NUM_CLASSES,
    activation=None
).to(DEVICE)

print("Architecture : Unet++")
print("Encoder      : EfficientNet-B3")
print("Input        : 1 channel")
print("Classes      :", NUM_CLASSES)
print("Parameters   :", f"{sum(p.numel() for p in model.parameters()):,}")


# ============================================================
# LOSS & METRICS
# ============================================================

class DiceLoss(nn.Module):
    def __init__(self, smooth=1.0):
        super().__init__()
        self.smooth = smooth

    def forward(self, logits, target):
        probabilities = torch.softmax(logits, dim=1)
        target_onehot = F.one_hot(target, num_classes=NUM_CLASSES).permute(0, 3, 1, 2).float()
        dims = (0, 2, 3)
        intersection = (probabilities * target_onehot).sum(dims)
        denominator = (probabilities + target_onehot).sum(dims)
        dice = (2 * intersection + self.smooth) / (denominator + self.smooth)
        return 1 - dice.mean()

dice_loss = DiceLoss()
cross_entropy = nn.CrossEntropyLoss()

def combined_loss(logits, target):
    return 0.6 * dice_loss(logits, target) + 0.4 * cross_entropy(logits, target)

def batch_metrics(logits, target):
    prediction = torch.argmax(logits, dim=1)
    dice_scores, iou_scores = [], []
    for cls in range(1, NUM_CLASSES):
        pred_cls = prediction == cls
        target_cls = target == cls
        intersection = (pred_cls & target_cls).sum().float()
        pred_area = pred_cls.sum().float()
        target_area = target_cls.sum().float()
        union = (pred_cls | target_cls).sum().float()
        dice = (2 * intersection + 1e-6) / (pred_area + target_area + 1e-6)
        iou = (intersection + 1e-6) / (union + 1e-6)
        dice_scores.append(dice.item())
        iou_scores.append(iou.item())
    return np.mean(dice_scores), np.mean(iou_scores)


# ============================================================
# OPTIMIZER & SCHEDULER
# ============================================================

optimizer = torch.optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
scaler = torch.amp.GradScaler("cuda")


# ============================================================
# TRAINING & VALIDATION FUNCTIONS
# ============================================================

def train_one_epoch():
    model.train()
    running_loss, running_dice = 0.0, 0.0
    progress = tqdm(train_loader, desc="Training", leave=False)
    for batch in progress:
        images = batch["image"].to(DEVICE, non_blocking=True)
        masks = batch["mask"].to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        with torch.autocast(device_type="cuda", dtype=torch.float16):
            logits = model(images)
            loss = combined_loss(logits, masks)
        scaler.scale(loss).backward()
        scaler.step(optimizer)
        scaler.update()
        dice, _ = batch_metrics(logits.detach(), masks)
        running_loss += loss.item()
        running_dice += dice
        progress.set_postfix(loss=f"{loss.item():.4f}", dice=f"{dice:.4f}")
    return running_loss / len(train_loader), running_dice / len(train_loader)

@torch.no_grad()
def validate():
    model.eval()
    running_loss, running_dice, running_iou = 0.0, 0.0, 0.0
    progress = tqdm(val_loader, desc="Validation", leave=False)
    for batch in progress:
        images = batch["image"].to(DEVICE, non_blocking=True)
        masks = batch["mask"].to(DEVICE, non_blocking=True)
        with torch.autocast(device_type="cuda", dtype=torch.float16):
            logits = model(images)
            loss = combined_loss(logits, masks)
        dice, iou = batch_metrics(logits, masks)
        running_loss += loss.item()
        running_dice += dice
        running_iou += iou
    return running_loss / len(val_loader), running_dice / len(val_loader), running_iou / len(val_loader)


# ============================================================
# TRAINING LOOP
# ============================================================

print("\n" + "=" * 70)
print("TRAINING START")
print("=" * 70)

history = []
best_val_dice = -np.inf
epochs_without_improvement = 0

for epoch in range(1, EPOCHS + 1):
    train_loss, train_dice = train_one_epoch()
    val_loss, val_dice, val_iou = validate()
    scheduler.step()
    current_lr = optimizer.param_groups[0]["lr"]
    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_dice": train_dice,
        "val_loss": val_loss,
        "val_dice": val_dice,
        "val_iou": val_iou,
        "lr": current_lr
    })
    print(f"\nEpoch {epoch:02d}/{EPOCHS}")
    print(f"Train Loss : {train_loss:.4f}")
    print(f"Train Dice : {train_dice:.4f}")
    print(f"Val Loss   : {val_loss:.4f}")
    print(f"Val Dice   : {val_dice:.4f}")
    print(f"Val IoU    : {val_iou:.4f}")
    print(f"LR         : {current_lr:.7f}")

    torch.save({
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "epoch": epoch,
        "val_dice": val_dice,
        "val_iou": val_iou,
        "image_size": IMAGE_SIZE,
        "num_classes": NUM_CLASSES,
        "architecture": "UnetPlusPlus",
        "encoder": "timm-efficientnet-b3"
    }, LAST_CHECKPOINT_PATH)

    if val_dice > best_val_dice:
        best_val_dice = val_dice
        epochs_without_improvement = 0
        torch.save({
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "epoch": epoch,
            "val_dice": val_dice,
            "val_iou": val_iou,
            "image_size": IMAGE_SIZE,
            "num_classes": NUM_CLASSES,
            "architecture": "UnetPlusPlus",
            "encoder": "timm-efficientnet-b3"
        }, CHECKPOINT_PATH)
        print("✓ Best checkpoint saved.")
    else:
        epochs_without_improvement += 1
        print(f"No improvement: {epochs_without_improvement}/{PATIENCE}")
        if epochs_without_improvement >= PATIENCE:
            print("\nEarly stopping.")
            break


# ============================================================
# TRAINING HISTORY & CURVES
# ============================================================

history_df = pd.DataFrame(history)
history_df.to_csv(OUTPUT_DIR / "training_history.csv", index=False)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
axes[0].plot(history_df["epoch"], history_df["train_loss"], label="Train")
axes[0].plot(history_df["epoch"], history_df["val_loss"], label="Validation")
axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()

axes[1].plot(history_df["epoch"], history_df["train_dice"], label="Train")
axes[1].plot(history_df["epoch"], history_df["val_dice"], label="Validation")
axes[1].set_title("Dice Score")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Dice")
axes[1].legend()
plt.tight_layout()
plt.show()


# ============================================================
# LOAD BEST CHECKPOINT
# ============================================================

print("\n" + "=" * 70)
print("CHECKPOINT VALIDATION")
print("=" * 70)

if not CHECKPOINT_PATH.exists():
    raise FileNotFoundError("Best checkpoint not found:", CHECKPOINT_PATH)

checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

print("Best epoch:", checkpoint["epoch"])
print("Best validation Dice:", f"{checkpoint['val_dice']:.4f}")
print("Best validation IoU:", f"{checkpoint['val_iou']:.4f}")


# ============================================================
# TEST EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST EVALUATION")
print("=" * 70)

test_dice_values, test_iou_values = [], []
with torch.no_grad():
    for batch in tqdm(test_loader, desc="Testing"):
        images = batch["image"].to(DEVICE)
        masks = batch["mask"].to(DEVICE)
        with torch.autocast(device_type="cuda", dtype=torch.float16):
            logits = model(images)
        dice, iou = batch_metrics(logits, masks)
        test_dice_values.append(dice)
        test_iou_values.append(iou)

test_dice = float(np.mean(test_dice_values))
test_iou = float(np.mean(test_iou_values))
print(f"\nTest Dice : {test_dice:.4f}")
print(f"Test IoU  : {test_iou:.4f}")


# ============================================================
# PER-CLASS TEST METRICS
# ============================================================

class_names = {1: "LV cavity", 2: "Myocardium", 3: "Left atrium"}
class_dice = {cls: [] for cls in class_names}

with torch.no_grad():
    for batch in tqdm(test_loader, desc="Per-class evaluation"):
        images = batch["image"].to(DEVICE)
        masks = batch["mask"].to(DEVICE)
        with torch.autocast(device_type="cuda", dtype=torch.float16):
            logits = model(images)
        predictions = torch.argmax(logits, dim=1)
        for cls in class_names:
            pred_cls = predictions == cls
            target_cls = masks == cls
            intersection = (pred_cls & target_cls).sum().float()
            denominator = pred_cls.sum() + target_cls.sum()
            dice = (2 * intersection + 1e-6) / (denominator + 1e-6)
            class_dice[cls].append(dice.item())

print("\nPer-class Dice:")
for cls, name in class_names.items():
    score = np.mean(class_dice[cls])
    print(f"{name:15s}: {score:.4f}")


# ============================================================
# SEGMENTATION XAI (Gradient-based Saliency)
# ============================================================

print("\n" + "=" * 70)
print("SEGMENTATION XAI (Gradient Saliency)")
print("=" * 70)

xai_row = test_df.iloc[0]
xai_image = nib.load(xai_row["image"]).get_fdata()
xai_gt = nib.load(xai_row["mask"]).get_fdata().astype(np.uint8)

xai_tensor = preprocess_image(xai_image).unsqueeze(0).to(DEVICE)
xai_tensor.requires_grad = True

model.zero_grad(set_to_none=True)
with torch.autocast(device_type="cuda", dtype=torch.float16):
    xai_logits = model(xai_tensor)
    xai_probabilities = torch.softmax(xai_logits.float(), dim=1)

# Target class: LV cavity (class 1)
TARGET_CLASS = 1
target_score = xai_probabilities[0, TARGET_CLASS].mean()
target_score.backward()

# Compute saliency map: absolute gradient of target score w.r.t. input
saliency = xai_tensor.grad.detach().abs()
saliency = saliency[0, 0].cpu().numpy()
saliency = (saliency - saliency.min()) / (saliency.max() + 1e-8)

# Prediction mask for overlay
prediction = torch.argmax(xai_probabilities, dim=1)[0].detach().cpu().numpy()
input_image = xai_tensor[0, 0].detach().cpu().numpy()

# Save XAI arrays
np.save(OUTPUT_DIR / "xai_lv_cavity_saliency.npy", saliency)
np.save(OUTPUT_DIR / "xai_prediction_mask.npy", prediction)


# ============================================================
# XAI VISUALIZATION
# ============================================================

fig, axes = plt.subplots(1, 5, figsize=(24, 5))

axes[0].imshow(input_image, cmap="gray")
axes[0].set_title("Echocardiography")

axes[1].imshow(xai_gt, interpolation="nearest")
axes[1].set_title("Ground Truth")

axes[2].imshow(prediction, interpolation="nearest")
axes[2].set_title("Predicted Segmentation")

axes[3].imshow(input_image, cmap="gray")
axes[3].imshow(saliency, alpha=0.45, cmap="jet")
axes[3].set_title("LV Saliency")

axes[4].imshow(input_image, cmap="gray")
axes[4].imshow(prediction, alpha=0.35)
axes[4].imshow(saliency, alpha=0.40, cmap="jet")
axes[4].set_title("Prediction + Saliency")

for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("CARDIOVISION AI | FINAL SUMMARY")
print("=" * 70)
print(f"Patients              : {len(patient_ids)}")
print(f"Train patients        : {len(train_patients)}")
print(f"Validation patients   : {len(val_patients)}")
print(f"Test patients         : {len(test_patients)}")
print(f"Train pairs           : {len(train_df)}")
print(f"Validation pairs      : {len(val_df)}")
print(f"Test pairs            : {len(test_df)}")
print(f"\nArchitecture          : Unet++")
print(f"Encoder               : EfficientNet-B3")
print(f"Input resolution      : {IMAGE_SIZE} × {IMAGE_SIZE}")
print(f"Classes               : {NUM_CLASSES}")
print(f"GPU                   : {GPU_NAME}")
print(f"\nBest validation Dice  : {checkpoint['val_dice']:.4f}")
print(f"Best validation IoU   : {checkpoint['val_iou']:.4f}")
print(f"Final test Dice       : {test_dice:.4f}")
print(f"Final test IoU        : {test_iou:.4f}")
print(f"\nCheckpoint (best): {CHECKPOINT_PATH}")
print(f"Checkpoint (last): {LAST_CHECKPOINT_PATH}")
print("\n✓ Raw CAMUS dataset used directly")
print("✓ Patient-level split")
print("✓ T4 GPU training (single device)")
print("✓ DataLoader workers disabled (num_workers=0)")
print("✓ Unet++ segmentation")
print("✓ Test evaluation")
print("✓ Segmentation XAI (gradient saliency)")
print("\nCardioVision ECHO model pipeline complete.")
```
